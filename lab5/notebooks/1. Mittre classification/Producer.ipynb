# Producer

Generates synthetic network/packet-like events and publishes them to the Kafka topic `events.raw`.

```python
import uuid
import json
import random
import time
from datetime import datetime
from kafka import KafkaProducer
```

```python
KAFKA_BROKER = "kafka:9092"
TOPIC = "events.raw"
NUM_EVENTS = 100

EVENT_TYPES = [
    "port_scan",
    "brute_force",
    "data_exfiltration",
    "lateral_movement",
    "c2_communication",
    "privilege_escalation",
    "normal_traffic",
]
```

```python
def generate_event():
    return {
        "event_id": str(uuid.uuid4()),
        "timestamp": datetime.utcnow().isoformat(),
        "src_ip": f"10.0.{random.randint(0, 255)}.{random.randint(1, 254)}",
        "dst_ip": f"192.168.{random.randint(0, 255)}.{random.randint(1, 254)}",
        "src_port": random.randint(1024, 65535),
        "dst_port": random.choice([22, 80, 443, 3389, 8080, 445, 3306]),
        "protocol": random.choice(["TCP", "UDP", "ICMP"]),
        "payload_size": random.randint(64, 65535),
        "event_type": random.choice(EVENT_TYPES),
    }
```

```python
producer = KafkaProducer(
    bootstrap_servers=KAFKA_BROKER,
    value_serializer=lambda v: json.dumps(v).encode("utf-8"),
)

for i in range(NUM_EVENTS):
    event = generate_event()
    producer.send(TOPIC, value=event)
    print(f"[{i+1}/{NUM_EVENTS}] Sent event: {event['event_id']} ({event['event_type']})")
    time.sleep(0.05)

producer.flush()
producer.close()
print(f"\nDone. {NUM_EVENTS} events published to topic '{TOPIC}'.")
```
